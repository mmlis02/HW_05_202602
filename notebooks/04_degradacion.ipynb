# 04 — Escalera de degradación

**Qué hace.** Genera los niveles L0–L3 de las llamadas de forma reproducible, mide la SNR real de L3 y obtiene las dos capas de métricas en cada nivel.

**Qué lee.** Los audios de llamadas (`rutas.audios_dir`), el ruido (`rutas.ruido_dir`), `config.yaml → degradacion` y `data/ground_truth/llamadas.csv`.

**Qué produce.** Los audios degradados en `rutas.audios_degradados_dir` (no se commitean), sus transcripciones en la caché y la curva de WER y exactitud por nivel en `rutas.resultados_dir`.

La transcripción de cada nivel necesita GPU: esa parte se corre en Colab, igual que el 02. Llama a `src.degradacion`, `src.engine` y `src.metricas`.

L0 tal como se grabó · L1 nota de voz de WhatsApp (compresión de bajo bitrate, mono) · L2 llamada telefónica (300–3400 Hz, 8 kHz) · L3 L1 + ruido ambiente.
